```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

# Download Microsoft stock data
data = yf.download("MSFT", start="2023-01-01", end="2025-01-01")

# Use Closing Price
price = data['Close']

# Remove missing values
price = price.dropna()

# Time Series Decomposition
result = seasonal_decompose(price, model='additive', period=30)

# -------------------------
# Original Stock Price
# -------------------------
plt.figure(figsize=(12,4))
plt.plot(price)
plt.title("Microsoft (MSFT) Original Stock Price")
plt.xlabel("Date")
plt.ylabel("Price")
plt.grid(True)
plt.savefig("msft_original.png")
plt.show()

print("=" * 150)

# -------------------------
# Trend
# -------------------------
plt.figure(figsize=(12,4))
plt.plot(result.trend, color='blue')
plt.title("Microsoft (MSFT) Trend Component")
plt.xlabel("Date")
plt.grid(True)
plt.savefig("msft_trend.png")
plt.show()

print("=" * 150)

# -------------------------
# Seasonal
# -------------------------
plt.figure(figsize=(12,4))
plt.plot(result.seasonal, color='green')
plt.title("Microsoft (MSFT) Seasonal Component")
plt.xlabel("Date")
plt.grid(True)
plt.savefig("msft_seasonal.png")
plt.show()

print("=" * 150)

# -------------------------
# Residual / Cyclical
# -------------------------
plt.figure(figsize=(12,4))
plt.plot(result.resid, color='red')
plt.title("Microsoft (MSFT) Residual (Cyclical)")
plt.xlabel("Date")
plt.grid(True)
plt.savefig("msft_cyclical.png")
plt.show()

# -------------------------
# Dynamic Result Analysis
# -------------------------
print("\n===== RESULT ANALYSIS =====")

if result.trend.iloc[-1] > result.trend.iloc[0]:
    print("Trend: The Microsoft stock shows an overall upward trend.")
else:
    print("Trend: The Microsoft stock shows an overall downward trend.")

if result.seasonal.abs().mean() > 1:
    print("Seasonal: Noticeable seasonal variations are present.")
else:
    print("Seasonal: Seasonal variations are weak.")

if result.resid.std() > 5:
    print("Residual: High irregular fluctuations are present.")
else:
    print("Residual: Low irregular fluctuations are present.")
```
